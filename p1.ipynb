```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

## a) Crea un vector de tiempo de 10 segundos de duración y una tasa de muestreo de 8,000 muestras por segundo.

Vector de tiempo = Frecuencia de muestreo * duración = ( 8000 ) * ( 10 ) = 80,000 muestras

```python
muestreo = 8000
duracion = 10
periodo_muestreo = 1 / muestreo
```

```python
t = np.arange( 0, duracion, periodo_muestreo )
print("Número total de muestras:", len( t ))
```

## b) Crea un vector de frecuencia angular w

$\omega = 2\pi f$

```python
frecuencias = np.array( [ 100, 20, 35 ] )
w = 2 * np.pi * frecuencias

print("Frecuencias angulares:", np.round( w, 2 ) )
```

## c) Crea una señal digital a partir de la función seno

Una señal senoidal puede representarse mediante: $x(t)=A\sin(\omega t+\phi)$

```python
inicio_medicion = time.perf_counter()

s1 = np.sin( w[0] * t )
s2 = 2 * np.sin( w[1] * t )
s3 = np.sin( w[2] * t )

s4 = s1 + s2 + s3

fin_medicion = time.perf_counter()
```

```python
plt.plot( t, s1 )

plt.xlim( 0, 0.05 )

plt.xlabel("Tiempo en s")
plt.ylabel("Amplitud")
plt.title("Señal senoidal")

plt.grid()
plt.show()
```

## d) Crea más señales digitales a partir de la función seno o la función coseno con amplitud o frecuencia diferentes

```python
plt.plot( t, s2, label="A = 2, f = 20 Hz", color="#F707D7", linestyle="--")
plt.plot( t, s3, label="A = 1, f = 30 Hz", color="#00E1FF" )

plt.xlim( 0, 0.09 )

plt.xlabel("Tiempo en s")
plt.ylabel("Amplitud")
plt.title("Señal senoidal a diferentes frecuencias")

plt.grid()
plt.legend()
plt.show()
```

## Suma todas las señales creadas y muestra una gráfica del resultado final

```python
plt.plot( t, s4 )

plt.xlim( 0, 0.1 )

plt.xlabel("Tiempo en s")
plt.ylabel("Amplitud")
plt.title("Señal senoidal a diferentes frecuencias")

plt.grid()
plt.show()
```

## f) Utiliza las funciones necesarias en cada entorno de programación para estimar el tiempo de procesamiento

Para estimar el tiempo requerido para la generación y procesamiento de las
señales se utiliza la función `perf_counter()` del módulo `time`.

El tiempo de procesamiento se obtiene mediante: $t_{procesamiento}=t_{final}-t_{inicial}$

```python
tiempo_procesamiento = np.round( ( fin_medicion - inicio_medicion ) * 1000 , 2 )

print("Tiempo de procesamiento:", tiempo_procesamiento, "ms")
```
